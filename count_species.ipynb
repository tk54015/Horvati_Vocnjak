# Fruit species counts from project data
This notebook counts the number of different fruit trees in the available JSON datasets, excluding vinova loza.

```python
import json
import os

# Put the notebook next to the data folder for relative access
base_path = os.path.join(os.path.dirname(__file__), 'data')
```

```python
# Define fruit counts by reading available stablo JSON export data
file_names = ['stabla.json', 'stabla-export-2026-04-16.json']
fruit_counts = {}
for file_name in file_names:
    path = os.path.join(base_path, file_name)
    with open(path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    items = data.get('items', [])
    if not items:
        items = data.get('stabla', [])
    if isinstance(items, dict):
        items = [items]

    for item in items:
        if not isinstance(item, dict):
            continue
        species = item.get('vrsta') or item.get('naziv')
        if not isinstance(species, str):
            continue
        species = species.strip().lower()
        fruit_counts[species] = fruit_counts.get(species, 0) + 1
```

```python
# Store counts in a cleaned dictionary and exclude vinova loza by default
cleaned_counts = {name: count for name, count in fruit_counts.items() if name != 'vinova_loza'}

# Normalize some known names for easier reading
normalized_counts = {
    'jabuka': cleaned_counts.get('jabuka', 0),
    'kruska': cleaned_counts.get('kruska', 0),
    'breskva': cleaned_counts.get('breskva', 0),
    'sljiva': cleaned_counts.get('sljiva', 0),
    'dunja': cleaned_counts.get('dunja', 0),
    'glog': cleaned_counts.get('glog', 0),
    'kupina': cleaned_counts.get('kupina', 0)
}
```

```python
# Compute total fruit count excluding vinova loza
selected_total = sum(normalized_counts.values())
selected_total
```

```python
# Display individual counts and totals
print('Individual fruit counts:')
for name, count in normalized_counts.items():
    print(f'- {name}: {count}')
print(f'\nTotal fruit count (excluding vinova loza): {selected_total}')
```
